```python
import pypsa
import pandas as pd

n = pypsa.Network("../networks/base.nc")

# --- LINKS: capacity that doesn't double-count directions
links = n.links.copy()
links["capacity"] = links["p_nom"].fillna(0.0)
links["corridor"] = links.apply(lambda r: (*sorted([r.bus0, r.bus1]), r.carrier), axis=1)

# effective corridor capacity = max over directions
links_eff = (links.groupby("corridor")["capacity"]
                  .max()
                  .rename("effective_corridor_capacity")
                  .reset_index())

# optional: cross-border aggregate by country pair (directionless)
if "country" in n.buses.columns:
    bus_cty = n.buses["country"]
    links["ctry_pair"] = links.apply(lambda r: tuple(sorted([bus_cty[r.bus0], bus_cty[r.bus1]])), axis=1)
    links_eff_country = (links.groupby(["corridor","ctry_pair"])["capacity"].max()
                               .groupby("ctry_pair").sum()
                               .sort_values(ascending=False))
    print(links_eff_country.head(20))

# --- LINES: avoid counting cloned entries
lines = n.lines.copy()
lines["capacity"] = lines["s_nom"].fillna(0.0)
lines["pair"] = lines.apply(
    lambda r: (*sorted([r.bus0, r.bus1]),
               r.carrier if "carrier" in lines.columns else "line",
               getattr(r, "type", ""),
               round(float(r.length) if hasattr(r, "length") else 0.0, 3)),
    axis=1
)

lines_eff = (lines.groupby("pair")["capacity"]
                  .max()
                  .rename("effective_corridor_capacity")
                  .reset_index())
```
